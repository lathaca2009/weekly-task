# Profit insights

# 1. Which category generates the highest profit?
 Ans: The Technology category generates the highest profit because many of its products have higher selling prices and profit margins.

# 2. Which sub-categories have low or negative profit?
 Ans: Some sub-categories, especially Tables and Bookcases, show low or negative profit because of high discounts and lower profit margins.

# 3. Which regions are most profitable?
 Ans:The West and East regions are among the most profitable regions, while the Central region generally has comparatively lower profit.

# Discount insights

# 1.Does profit decrease as discount increases?
   Ans: Yes. In general, when the discount increases, the profit tends to decrease because the store earns less from the sale.

# 2.Which discount levels have poor profitability?
 Ans: Higher discount levels usually have poor profitability. Large discounts can reduce the profit margin and may even lead to losses.

# 3.Which high-discount transactions generate negative profit?
  Ans: Some products sold with very high discounts generate negative profit. This means the discount given is too high compared 
       with the profit earned from the product.

# Correlation insights

#  1.Which variables have strong positive correlation?
   Ans:Sales and Profit generally show a positive relationship. When sales increase, profit also tends to increase, 
       although this may vary depending on the discount and product.
 
# 2.Which variables have negative correlation?
   Ans:Discount and Profit have a negative relationship. As the discount increases, the profit generally decreases.

# 3.Is Discount negatively related to Profit?
   Ans:Yes. The analysis suggests that higher discounts are associated with lower profit. Therefore, the store should carefully
